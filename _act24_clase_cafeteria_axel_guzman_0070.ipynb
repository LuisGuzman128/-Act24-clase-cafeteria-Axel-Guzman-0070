```python
# Act24 clase Cafetera Axel Guzman 0070
# Programa interactivo para el registro y uso de 2 cafeteras

class Cafetera:
    def __init__(self, marca, capacidadLitros, tipoFiltro, potencia):
        self.marca = marca
        self.capacidadLitros = capacidadLitros
        self.tipoFiltro = tipoFiltro
        self.potencia = potencia
        self.temperatura_lista = False

    # Funciones de la clase
    def calentarAgua(self):
        self.temperatura_lista = True
        print(f"🔥 [SISTEMA]: Calentando agua en la cafetera {self.marca} ({self.potencia}W)...")

    def molerGranos(self):
        print(f"☕ [SISTEMA]: Moliendo granos para filtro tipo {self.tipoFiltro}.")

    def servirTaza(self):
        if self.temperatura_lista:
            print(f"☕ [SISTEMA]: Taza servida. ¡Disfruta tu café de la marca {self.marca}!")
        else:
            print(f"⚠️ [ERROR]: El agua de la cafetera {self.marca} aún está fría.")

# --- BLOQUE INTERACTIVO: REGISTRO DE DATOS ---

# Registro de la Primera Cafetera
print("=== REGISTRO CAFETERA 1 ===")
m1 = input("Marca: ")
c1 = input("Capacidad (Litros): ")
f1 = input("Tipo de filtro: ")
p1 = input("Potencia (Watts): ")
cafetera1 = Cafetera(m1, c1, f1, p1)

# Registro de la Segunda Cafetera
print("\n=== REGISTRO CAFETERA 2 ===")
m2 = input("Marca: ")
c2 = input("Capacidad (Litros): ")
f2 = input("Tipo de filtro: ")
p2 = input("Potencia (Watts): ")
cafetera2 = Cafetera(m2, c2, f2, p2)

# --- EJECUCIÓN DE RESULTADOS ---

print("\n--- PROCESANDO CAFETERA 1 ---")
cafetera1.molerGranos()
cafetera1.calentarAgua()
cafetera1.servirTaza()

print("\n--- PROCESANDO CAFETERA 2 ---")
cafetera2.molerGranos()
# En esta no calentamos agua para probar la validación
cafetera2.servirTaza()

print("\n-----------------------")
print("Programa terminado")
print("creador de software por Axel Guzman NL= 0070")
```
